## 1. Basic Function

Use `def` to define a reusable function.

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Ram"))
```

## 2. Mean Function

```python
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)

marks = [78, 85, 92, 66, 88]
print("Mean:", calculate_mean(marks))
```

## 3. Maximum Function

```python
def find_maximum(numbers):
    return max(numbers)

print("Maximum:", find_maximum(marks))
```

## 4. Minimum Function

```python
def find_minimum(numbers):
    return min(numbers)

print("Minimum:", find_minimum(marks))
```

## 5. Percentage Function

This example demonstrates parameters and a returned value.

```python
def calculate_percentage(obtained, total):
    return (obtained / total) * 100

percentage = calculate_percentage(425, 500)
print("Percentage:", percentage, "%")
```

## 6. Grade Function

```python
def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

print("Grade:", calculate_grade(85))
```

## 7. Student Summary Function

Combine reusable functions to summarize student marks.

```python
def student_summary(name, marks):
    total = sum(marks)
    percentage = calculate_percentage(total, len(marks) * 100)
    return {
        "Name": name,
        "Mean": calculate_mean(marks),
        "Maximum": find_maximum(marks),
        "Minimum": find_minimum(marks),
        "Percentage": percentage,
        "Grade": calculate_grade(percentage)
    }

summary = student_summary("Ram", [78, 85, 92, 66, 88])
for key, value in summary.items():
    print(f"{key}: {value}")
```

## 8. Employee Salary Functions

```python
def calculate_annual_salary(monthly_salary):
    return monthly_salary * 12

def calculate_bonus(salary, bonus_rate=10):
    return salary * bonus_rate / 100

monthly_salary = 30000
annual_salary = calculate_annual_salary(monthly_salary)
print("Annual Salary:", annual_salary)
print("10% Bonus:", calculate_bonus(annual_salary))
```

## 9. Parameters and Return Values

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 20)
print("Returned value:", result)
```

## 10. Mini Practice Task

Create reusable functions for total, range, values above a threshold, discount amount, and Celsius-to-Fahrenheit conversion.

```python
def calculate_total(numbers):
    return sum(numbers)

def calculate_range(numbers):
    return max(numbers) - min(numbers)

def count_above(numbers, threshold):
    return sum(1 for number in numbers if number > threshold)

def calculate_discount(price, discount_rate):
    return price * discount_rate / 100

def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

values = [10, 25, 40, 55, 70]
print("Total:", calculate_total(values))
print("Range:", calculate_range(values))
print("Above 40:", count_above(values, 40))
print("10% Discount:", calculate_discount(1000, 10))
print("25°C in Fahrenheit:", celsius_to_fahrenheit(25))
```

## Conclusion

Functions make Python code easier to organize, reuse, test, and maintain. In data science, reusable functions are useful for repeated calculations, transformations, and preprocessing steps.
