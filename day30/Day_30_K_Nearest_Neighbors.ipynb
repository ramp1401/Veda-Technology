**Veda Technology · Level 2 · Day 30 · Data Science Track**

Build a KNN classification model and evaluate how different values of K influence prediction performance.

**Objective:** Understand distance-based classification and the importance of feature scaling.

**Tools:** Python, Pandas, NumPy, Scikit-learn, Matplotlib, Jupyter Notebook

**Dataset:** Iris Dataset


## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## 2. Load and Inspect Data

```python
iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = pd.Series(iris.target, name="target")
print("Shape:", X.shape)
print("Classes:", iris.target_names)
print("Missing values:", X.isnull().sum().sum())
X.head()
```

## 3. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 4. Feature Scaling

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print("Scaled training shape:", X_train_scaled.shape)
```

## 5. KNN Model with K=5

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)
pred = knn.predict(X_test_scaled)
print("K=5 test accuracy:", round(accuracy_score(y_test, pred), 4))
```

## 6. Test Multiple K Values

```python
results = []
for k in range(1, 21):
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_scaled, y_train)
    train_pred = model.predict(X_train_scaled)
    test_pred = model.predict(X_test_scaled)
    results.append({
        "K": k,
        "Training Accuracy": accuracy_score(y_train, train_pred),
        "Testing Accuracy": accuracy_score(y_test, test_pred)
    })
results_df = pd.DataFrame(results)
results_df.round(4)
```

```python
plt.figure(figsize=(9,5))
plt.plot(results_df["K"], results_df["Training Accuracy"], marker="o", label="Training")
plt.plot(results_df["K"], results_df["Testing Accuracy"], marker="o", label="Testing")
plt.xlabel("K")
plt.ylabel("Accuracy")
plt.title("KNN Performance for Different K Values")
plt.xticks(range(1,21))
plt.ylim(0.7,1.05)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 7. Best K Selection

```python
best_row = results_df.loc[results_df["Testing Accuracy"].idxmax()]
best_k = int(best_row["K"])
print("Best K:", best_k)
print("Best testing accuracy:", round(best_row["Testing Accuracy"], 4))
```

## 8. Final Model Evaluation

```python
final_model = KNeighborsClassifier(n_neighbors=best_k)
final_model.fit(X_train_scaled, y_train)
final_pred = final_model.predict(X_test_scaled)

print(classification_report(y_test, final_pred, target_names=iris.target_names))
print("Confusion Matrix:")
print(confusion_matrix(y_test, final_pred))
```

## 9. Interpretation

- **K too small:** the model can be sensitive to noise and overfit.
- **K too large:** the model can become too smooth and underfit.
- **Scaling:** important because KNN uses distances between observations.
- The best K should be selected using validation/test performance rather than chosen arbitrarily.


## 10. Interview Questions

1. **How does KNN classify observations?** It finds the K closest training observations and uses the majority class.
2. **Why is scaling important for KNN?** Because KNN is distance-based and large-scale features can dominate distance calculations.
3. **What happens when K is too small?** The model may overfit and become sensitive to noise.
4. **What happens when K is too large?** The model may underfit.
5. **Is KNN parametric or non-parametric?** KNN is generally considered non-parametric.
