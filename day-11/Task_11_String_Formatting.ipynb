## 1. Quick Reference & Core Concepts


| Technique | Syntax Example | Data Science Use Case |
| :--- | :--- | :--- |
| f-Strings (Python 3.6+) | f"{metric:.2f}" | High-performance logging, automated report generation |
| str.format() | "{:04d}".format(id) | Template reusability and dynamic SQL/query construction |
| Precision & Specifiers | f"{val:,.2f}" | Financial reporting ($1,234.50) and percentage formatting |
| Alignment & Padding | f"{col:<15} | {val:>8}" | ASCII tabular dashboards and terminal console outputs |
| Text Methods | .strip(), .lower(), .replace() | Feature standardization and raw string sanitization |

## 2. Practical Code Examples

```python
# Example 1: f-Strings with Numerical Precision and Currency Formatting
revenue = 1458920.4567
growth_pct = 0.14285
print(f"Ex 1 - Formatted Revenue: ${revenue:,.2f} | Growth Rate: {growth_pct:.2%}")
```

```python
# Example 2: Text Alignment, Padding, and Tabular Widths
headers = ["Feature", "Importance", "Type"]
print(f"{headers[0]:<15} | {headers[1]:>10} | {headers[2]:^10}")
print("-" * 42)
print(f"{'credit_score':<15} | {0.3542:>10.4f} | {'Float':^10}")
print(f"{'age':<15} | {0.2189:>10.4f} | {'Integer':^10}")
```

```python
# Example 3: Reusable Templates with str.format()
query_template = "SELECT {cols} FROM {table} WHERE {cond} = '{val}';"
user_query = query_template.format(cols="id, score, attendance", table="students", cond="cohort", val="2026-A")
print(f"Ex 3 - Constructed Query:
  {user_query}")
```

```python
# Example 4: Padding Numbers with Leading Zeros (ID Standardization)
raw_ids = [5, 42, 109, 3421]
standard_ids = [f"EMP-{i:05d}" for i in raw_ids]
print(f"Ex 4 - Standardized IDs: {standard_ids}")
```

```python
# Example 5: String Preprocessing (strip, lower, replace, split)
dirty_titles = ["  SENIOR DATA SCIENTIST  ", "Machine_learning_engineer", "bi_analyst "]
clean_titles = [t.strip().replace("_", " ").title() for t in dirty_titles]
print(f"Ex 5 - Cleaned Titles: {clean_titles}")
```

## 3. Applied Data Science Pipeline
### Applied Data Pipeline: Automated Student Performance & Payroll Text Report Generator

```python
students = [
    {"id": 101, "name": "Aman", "score": 88.456, "status": "Pass"},
    {"id": 102, "name": "Pooja", "score": 92.1, "status": "Distinction"},
    {"id": 103, "name": "Rohan", "score": 45.8, "status": "Remedial"}
]
print("=" * 48)
print(f"{'ID':^6} | {'Student Name':<12} | {'Score (%)':>10} | {'Status':^12}")
print("=" * 48)
for s in students:
    print(f"{s['id']:04d}   | {s['name']:<12} | {s['score']:>10.2f} | {s['status']:^12}")
print("=" * 48)
```
