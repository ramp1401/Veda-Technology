## 1. Syntax Comparison: Traditional Loops vs Comprehensions

Comprehensions provide a more readable, declarative, and computationally optimized alternative to explicit accumulator loops.

```python
# Traditional for loop vs List Comprehension
numbers = [1, 2, 3, 4, 5]

# Traditional Loop approach
squares_loop = []
for num in numbers:
    squares_loop.append(num ** 2)

# List Comprehension approach
squares_comp = [num ** 2 for num in numbers]

print(f"Original Numbers: {numbers}")
print(f"Squares via Loop: {squares_loop}")
print(f"Squares via Comp: {squares_comp}")
```

## 2. List Comprehension Examples (5 Scenarios)
Covering basic mapping, filtering, string cleaning, conditional assignments (`if-else`), and flattened matrices.

```python
# Example 1: Basic Numerical Transformation (Square Roots / Scaling)
# Converting raw test scores to percentages out of 100
raw_marks = [35, 42, 48, 28, 50]  # out of 50
percentages = [(score / 50) * 100 for score in raw_marks]
print(f"Ex 1 - Percentages: {percentages}")
```

```python
# Example 2: Data Filtering with 'if' condition
# Filter out passing marks (threshold >= 70%)
passing_marks = [score for score in percentages if score >= 70]
print(f"Ex 2 - Passing Marks (>= 70%): {passing_marks}")
```

```python
# Example 3: Text Preprocessing & Cleaning
# Strip whitespace and standardize category names to lowercase
raw_categories = ["  Electronics ", "Home & KITCHEN", "  books", "BEAUTY  "]
cleaned_categories = [cat.strip().lower() for cat in raw_categories]
print(f"Ex 3 - Cleaned Categories: {cleaned_categories}")
```

```python
# Example 4: Conditional Labeling using ternary 'if-else'
# Classify temperatures into 'High' or 'Normal'
temperatures = [31.5, 38.2, 35.0, 42.1, 29.8]
temp_labels = ["Heatwave" if temp >= 38.0 else "Moderate" for temp in temperatures]
print(f"Ex 4 - Temperature Labels: {temp_labels}")
```

```python
# Example 5: Flattening a 2D Matrix / Multi-feature List
# Flatten nested batch survey responses into a single list
batch_scores = [[85, 90], [78, 88], [92, 95]]
flattened_scores = [score for batch in batch_scores for score in batch]
print(f"Ex 5 - Flattened Scores: {flattened_scores}")
```

## 3. Dictionary Comprehension Examples (3+ Scenarios)
Covering mapping, condition filtering, and key inversion/aggregation.

```python
# Example 1: Mapping Lists to a Dictionary (Key-Value Pair Generation)
# Map student names to their raw scores
students = ["Aman", "Pooja", "Rohan", "Sneha", "Vikas"]
marks = [88, 72, 45, 95, 62]

student_marks_dict = {student: mark for student, mark in zip(students, marks)}
print(f"Dict Ex 1 - Student Marks Map: {student_marks_dict}")
```

```python
# Example 2: Filtering Dictionary Records with 'if'
# Filter products that are low in inventory (stock < 15)
inventory = {"Laptop": 25, "Mouse": 8, "Keyboard": 12, "Monitor": 18, "Headphones": 5}
low_stock = {product: count for product, count in inventory.items() if count < 15}
print(f"Dict Ex 2 - Low Stock Alerts: {low_stock}")
```

```python
# Example 3: Value Transformation with Conditional Classification
# Categorize product price tiers: 'Premium' if >= 1000 else 'Budget'
product_prices = {"Laptop": 1200, "Mouse": 25, "Keyboard": 60, "Monitor": 250, "Workstation": 2400}
price_tiers = {item: ("Premium" if price >= 500 else "Standard") for item, price in product_prices.items()}
print(f"Dict Ex 3 - Product Price Tiers: {price_tiers}")
```

```python
# Example 4: Currency Conversion & Key Inversion
# Convert product USD prices to INR (assuming 1 USD = 83 INR)
usd_prices = {"item_A": 10, "item_B": 45, "item_C": 120}
inr_prices = {item: round(usd * 83, 2) for item, usd in usd_prices.items()}
print(f"Dict Ex 4 - INR Converted Prices: {inr_prices}")
```

## 4. End-to-End Applied Pipelines on Suggested Datasets
Demonstrating end-to-end filtering and transformation workflows using **Student Marks Dataset** and **Product Dataset**.

```python
# Pipeline 1: Student Marks Preprocessing & Evaluation Pipeline
student_records = [
    {"id": 101, "name": "Aman", "marks": [78, 85, 92], "attendance": 88},
    {"id": 102, "name": "Pooja", "marks": [55, 62, 58], "attendance": 65},
    {"id": 103, "name": "Rohan", "marks": [40, 35, 45], "attendance": 80},
    {"id": 104, "name": "Sneha", "marks": [92, 95, 98], "attendance": 95},
    {"id": 105, "name": "Vikas", "marks": [82, 79, 85], "attendance": 72}
]

# 1. List Comprehension: Compute average mark per student
student_averages = [
    {"name": s["name"], "avg_score": round(sum(s["marks"]) / len(s["marks"]), 1), "attendance": s["attendance"]}
    for s in student_records
]

# 2. Dictionary Comprehension: Honor roll students (Avg >= 80 AND Attendance >= 75)
honor_roll = {
    s["name"]: s["avg_score"]
    for s in student_averages
    if s["avg_score"] >= 80 and s["attendance"] >= 75
}

print("=== STUDENT EVALUATION PIPELINE REPORT ===")
print(f"Processed Student Averages: {student_averages}")
print(f"\nHonor Roll Qualified Students: {honor_roll}")
```

```python
# Pipeline 2: Product Catalog Filtering & Discount Transformer
product_catalog = [
    {"sku": "P01", "title": " wireless gaming mouse ", "category": "Peripherals", "price": 45.0, "in_stock": True},
    {"sku": "P02", "title": "Mechanical Keyboard", "category": "Peripherals", "price": 120.0, "in_stock": False},
    {"sku": "P03", "title": " 4K IPS Monitor ", "category": "Displays", "price": 350.0, "in_stock": True},
    {"sku": "P04", "title": "USB-C Hub  ", "category": "Accessories", "price": 30.0, "in_stock": True},
    {"sku": "P05", "title": "Ergonomic Chair", "category": "Furniture", "price": 280.0, "in_stock": True}
]

# 1. List Comprehension: Filter only in-stock items and clean text titles
active_products = [
    {"sku": p["sku"], "title": p["title"].strip().title(), "price": p["price"], "category": p["category"]}
    for p in product_catalog
    if p["in_stock"]
]

# 2. Dictionary Comprehension: Apply 10% discount on Peripherals and Displays
discounted_pricing = {
    p["title"]: round(p["price"] * 0.90, 2) if p["category"] in ["Peripherals", "Displays"] else p["price"]
    for p in active_products
}

print("\n=== PRODUCT CATALOG PREPROCESSING REPORT ===")
print("Active In-Stock Products:")
for item in active_products:
    print(f"  [{item['sku']}] {item['title']:<25} | Base Price: ${item['price']}")

print("\nDiscounted Pricing Table:")
for product_name, final_price in discounted_pricing.items():
    print(f"  {product_name:<25} -> ${final_price:.2f}")
```
