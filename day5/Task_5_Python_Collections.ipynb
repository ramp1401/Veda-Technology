## 1. Comparison Table: Python Collection Types

| Collection | Ordered | Mutable | Duplicates Allowed | Syntax | Primary Use Case in Data Science |
| :--- | :---: | :---: | :---: | :---: | :--- |
| **List** | Yes | Yes | Yes | `[item1, item2]` | Dynamic sequences, feature lists, data pipelines |
| **Tuple** | Yes | No | Yes | `(item1, item2)` | Immutable metadata, fixed dimensions (e.g. `df.shape`), coordinates |
| **Set** | No | Yes | No | `{item1, item2}` | Deduplication, membership testing, set operations (union/intersection) |
| **Dictionary** | Yes (3.7+) | Yes | No (Keys) / Yes (Values) | `{"key": value}` | Key-value mapping, structured records, JSON/API payloads |

## 2. Python Lists: Operations & Methods
Covers `append()`, `remove()`, `sort()`, indexing, searching, and updating elements.

```python
# Initialize a list of student marks
student_marks = [78, 92, 65, 84, 55]
print(f"Initial Marks: {student_marks}")

# 1. Adding an element (append)
student_marks.append(88)
print(f"After append(88): {student_marks}")

# 2. Updating an element via index
student_marks[2] = 70
print(f"After updating index 2 to 70: {student_marks}")

# 3. Removing an element (remove)
student_marks.remove(55)
print(f"After remove(55): {student_marks}")

# 4. Searching for an element using 'in' and index()
search_val = 84
if search_val in student_marks:
    pos = student_marks.index(search_val)
    print(f"Found {search_val} at index {pos}")

# 5. Sorting elements (sort)
student_marks.sort()
print(f"After sort() ascending: {student_marks}")
student_marks.sort(reverse=True)
print(f"After sort(reverse=True): {student_marks}")
```

## 3. Python Tuples: Immutable Structures
Demonstrating index access, packing/unpacking, and immutability guarantees.

```python
# Define an immutable employee metadata record: (ID, Department, Joining Year)
employee_meta = (1001, "Data Science", 2024)
print(f"Employee Tuple: {employee_meta}")

# Accessing elements by index
print(f"Employee ID: {employee_meta[0]}")
print(f"Department: {employee_meta[1]}")

# Tuple unpacking
emp_id, dept, year = employee_meta
print(f"Unpacked: ID={emp_id}, Dept={dept}, Year={year}")

# Immutability demonstration (handling attempt to alter tuple)
try:
    employee_meta[1] = "Engineering"  # Raises TypeError
except TypeError as e:
    print(f"Immutability verified (TypeError caught): {e}")
```

## 4. Python Sets: Unique Collections & Set Algebra
Covers deduplication, `add()`, `discard()`, `union()`, `intersection()`, and `difference()`.

```python
# Deduplication: converting raw skill tags from a survey to a set
raw_skills = ["Python", "SQL", "Python", "Machine Learning", "SQL", "Power BI"]
unique_skills = set(raw_skills)
print(f"Raw list with duplicates: {raw_skills}")
print(f"Deduplicated Set: {unique_skills}")

# Adding and removing elements
unique_skills.add("Deep Learning")
unique_skills.discard("Power BI")
print(f"Updated Set: {unique_skills}")

# Set Algebra for skill gap analysis
required_job_skills = {"Python", "SQL", "TensorFlow", "Docker", "Machine Learning"}
candidate_skills = {"Python", "SQL", "Power BI", "Excel"}

matching_skills = candidate_skills.intersection(required_job_skills)
missing_skills = required_job_skills.difference(candidate_skills)
combined_portfolio = candidate_skills.union(required_job_skills)

print(f"Matching Skills: {matching_skills}")
print(f"Missing Skills (Skill Gap): {missing_skills}")
print(f"Total Unique Combined Skills: {combined_portfolio}")
```

## 5. Python Dictionaries: Key-Value Management
Demonstrating key lookup, `get()`, updating, removing (`pop()`), and iteration (`keys()`, `values()`, `items()`).

```python
# Structured employee profile dictionary
employee = {
    "id": "E-204",
    "name": "Aman Verma",
    "role": "Data Analyst",
    "base_salary": 65000
}
print(f"Initial Record: {employee}")

# Accessing values safely using .get()
print(f"Employee Name: {employee['name']}")
print(f"Department (with fallback): {employee.get('department', 'Unassigned')}")

# Updating existing value and adding a new key-value pair
employee["base_salary"] = 72000
employee["city"] = "New Delhi"
print(f"After Update: {employee}")

# Removing a key using pop()
removed_city = employee.pop("city")
print(f"Removed Key 'city' ({removed_city}): {employee}")

# Iterating over items
print("\n--- Record Fields ---")
for key, val in employee.items():
    print(f"{key:>12}: {val}")
```

## 6. End-to-End Applied Programs (Using Suggested Datasets)
Applying combined collection techniques to process realistic **Student** and **Employee** data.

```python
# Program 1: Student Gradebook Processing
# Dataset: List of dictionaries representing student academic records
student_dataset = [
    {"id": 101, "name": "Rohan", "scores": [85, 90, 78], "subjects": {"Math", "Physics", "CS"}},
    {"id": 102, "name": "Pooja", "scores": [60, 58, 64], "subjects": {"Math", "Chemistry", "CS"}},
    {"id": 103, "name": "Sneha", "scores": [92, 95, 96], "subjects": {"Math", "Physics", "CS"}},
    {"id": 104, "name": "Vikas", "scores": [45, 52, 48], "subjects": {"Biology", "Chemistry"}}
]

print("=== STUDENT PERFORMANCE & SUBJECT AUDIT ===")
all_enrolled_subjects = set()

for s in student_dataset:
    avg_score = round(sum(s["scores"]) / len(s["scores"]), 2)
    s["average_score"] = avg_score
    
    # Union to collect all unique subjects across the cohort
    all_enrolled_subjects = all_enrolled_subjects.union(s["subjects"])
    
    print(f"ID: {s['id']} | Name: {s['name']:<6} | Avg Score: {avg_score:>5.2f} | Subjects: {sorted(list(s['subjects']))}")

print(f"\nTotal Unique Subjects Offered Across Cohort: {all_enrolled_subjects}")
```

```python
# Program 2: Employee Department & Compensation Aggregator
# Dataset: List of dictionaries with tuple coordinates and skill sets
employee_dataset = [
    {"emp_id": "E1", "name": "Aditi", "dept": "Tech", "salary": 110000, "skills": {"Python", "SQL", "GCP"}},
    {"emp_id": "E2", "name": "Karan", "dept": "Sales", "salary": 48000, "skills": {"CRM", "Negotiation"}},
    {"emp_id": "E3", "name": "Neha", "dept": "Tech", "salary": 85000, "skills": {"Python", "Docker", "SQL"}},
    {"emp_id": "E4", "name": "Rahul", "dept": "HR", "salary": 52000, "skills": {"Recruiting", "Payroll"}}
]

# Departmental salary aggregation and tech skill registry
dept_salaries = {}
tech_skills = set()

for emp in employee_dataset:
    d = emp["dept"]
    dept_salaries[d] = dept_salaries.get(d, 0) + emp["salary"]
    
    if d == "Tech":
        tech_skills = tech_skills.union(emp["skills"])

print("=== DEPARTMENTAL SPEND & TECH SKILLS REPORT ===")
for dept_name, total_spend in dept_salaries.items():
    print(f"Department: {dept_name:<6} | Total Payroll: ${total_spend:,}")

print(f"\nTech Team Master Skill Portfolio: {tech_skills}")
```
