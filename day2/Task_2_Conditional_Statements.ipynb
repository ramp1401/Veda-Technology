## Part 1: Core Conditional Examples (10 Scenarios)

```python
# Example 1: Basic Numerical Comparison (> / >=)
# Check whether a student passed a test
score = 75
passing_mark = 40

if score >= passing_mark:
    print(f"Result: Passed with score {score}")
else:
    print(f"Result: Failed with score {score}")
```

```python
# Example 2: Categorical Exact Match (==)
# Check employee department assignment
department = "Analytics"

if department == "Analytics":
    print("Assigned to Data Science & Analytics Team")
else:
    print("Assigned to General Operations")
```

```python
# Example 3: Multi-tier Categorization using elif
# Categorize employee salary bands (Low / Medium / High)
salary = 65000

if salary >= 100000:
    band = "High"
elif salary >= 50000:
    band = "Medium"
else:
    band = "Low"

print(f"Salary: ${salary} -> Band: {band}")
```

```python
# Example 4: Logical 'and' Operator (Both conditions must be true)
# Student eligibility: Score >= 75 AND attendance >= 75%
score = 82
attendance_percentage = 80

if score >= 75 and attendance_percentage >= 75:
    print("Eligible for Honors Certification")
else:
    print("Not eligible for Honors Certification")
```

```python
# Example 5: Logical 'or' Operator (At least one condition true)
# Bonus eligibility: High rating OR 5+ years experience
performance_rating = "Excellent"
years_of_experience = 2

if performance_rating == "Excellent" or years_of_experience >= 5:
    print("Eligible for Annual Incentive/Bonus")
else:
    print("Standard annual review")
```

```python
# Example 6: Logical 'not' Operator
# Flagging unverified user profiles
is_verified = False

if not is_verified:
    print("Flag: User profile requires manual verification.")
else:
    print("Profile verified.")
```

```python
# Example 7: Nested Conditions
# Scholarship evaluation: must be enrolled, then check GPA
is_enrolled = True
gpa = 3.8

if is_enrolled:
    if gpa >= 3.5:
        print("Scholarship: Full Merit Scholarship Approved")
    else:
        print("Scholarship: Standard Grant")
else:
    print("Status: Candidate is not currently enrolled")
```

```python
# Example 8: Membership / Categorical Inclusion (in operator)
# Check city tier for logistics & operations
city = "New Delhi"
tier_1_cities = ["New Delhi", "Mumbai", "Bengaluru", "Chennai", "Hyderabad"]

if city in tier_1_cities:
    print(f"{city} is classified as a Tier-1 Metropolitan Hub")
else:
    print(f"{city} is classified as a Regional/Tier-2 Center")
```

```python
# Example 9: Chained Range Comparison
# Check if employee weekly work hours fall in standard range
weekly_hours = 42

if 35 <= weekly_hours <= 45:
    print("Working hours within standard range")
elif weekly_hours > 45:
    print("Overtime logged")
else:
    print("Under hours logged")
```

```python
# Example 10: Handling Empty/Default Values
# Validate input comments or remarks
remarks = ""

if remarks.strip():
    print(f"Feedback logged: {remarks}")
else:
    print("No feedback provided (assigned default 'N/A')")
```

## Part 2: Applied Classification Programs
Demonstrating end-to-end data classification on the suggested datasets:
1. **Student Performance Dataset**
2. **Employee Salary Dataset**

```python
# Classification Program 1: Student Performance Dataset
students = [
    {"name": "Aman", "score": 88, "attendance": 90},
    {"name": "Pooja", "score": 55, "attendance": 65},
    {"name": "Rohan", "score": 38, "attendance": 80},
    {"name": "Sneha", "score": 74, "attendance": 92},
    {"name": "Vikas", "score": 91, "attendance": 58}
]

print("=== STUDENT PERFORMANCE CLASSIFICATION REPORT ===")
for s in students:
    score = s["score"]
    att = s["attendance"]
    
    # Multi-condition classification logic
    if score >= 80 and att >= 75:
        status = "Distinction"
    elif score >= 50 and att >= 70:
        status = "Pass"
    elif score >= 50 and att < 70:
        status = "Conditional Pass (Attendance Warning)"
    else:
        status = "Fail"
        
    print(f"Student: {s['name']:<7} | Score: {score:>3} | Attendance: {att:>2}% | Status: {status}")
```

```python
# Classification Program 2: Employee Salary & Appraisal Dataset
employees = [
    {"emp_id": 101, "name": "Aditi", "dept": "Tech", "salary": 115000, "rating": 4.8},
    {"emp_id": 102, "name": "Karan", "dept": "Sales", "salary": 42000, "rating": 3.5},
    {"emp_id": 103, "name": "Neha", "dept": "Tech", "salary": 78000, "rating": 4.2},
    {"emp_id": 104, "name": "Rahul", "dept": "HR", "salary": 48000, "rating": 4.6}
]

print("=== EMPLOYEE COMPENSATION & APPRAISAL REPORT ===")
for emp in employees:
    sal = emp["salary"]
    rating = emp["rating"]
    
    # Salary bracket classification
    if sal >= 100000:
        salary_tier = "Executive / Senior"
    elif sal >= 50000:
        salary_tier = "Mid-Level"
    else:
        salary_tier = "Entry-Level"
        
    # Promotion / hike eligibility
    if rating >= 4.5 or (salary_tier == "Entry-Level" and rating >= 4.0):
        appraisal = "High Hike & Bonus Eligible"
    elif rating >= 3.5:
        appraisal = "Standard Hike"
    else:
        appraisal = "Performance Improvement Plan"
        
    print(f"ID: {emp['emp_id']} | Name: {emp['name']:<6} | Dept: {emp['dept']:<5} | Tier: {salary_tier:<18} | Action: {appraisal}")
```
