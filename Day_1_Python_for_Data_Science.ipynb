## 1. Variables

A variable is a name used to store a value.

Example:
```python
age = 22
salary = 35000
name = "Ram"
```

Python automatically determines the type of the value.

````python
# Creating variables
name = "Ram"
age = 22
salary = 35000
score = 87.5
is_student = True

print("Name:", name)
print("Age:", age)
print("Salary:", salary)
print("Score:", score)
print("Student:", is_student)
````

## 2. Main Python Data Types

| Data type | Example | Meaning |
|---|---|---|
| `int` | `22` | Whole number |
| `float` | `87.5` | Decimal number |
| `str` | `"Ram"` | Text |
| `bool` | `True` / `False` | Logical value |
| `list` | `[80, 90, 75]` | Ordered, changeable collection |
| `tuple` | `(80, 90, 75)` | Ordered, unchangeable collection |
| `set` | `{80, 90, 75}` | Unique values |
| `dict` | `{"name": "Ram"}` | Key-value pairs |


````python
# Integer
age = 22

# Float
percentage = 87.5

# String
student_name = "Ram"

# Boolean
passed = True

# List
scores = [85, 90, 78, 92]

# Tuple
coordinates = (19.07, 72.87)

# Set
unique_scores = {85, 90, 78, 92, 85}

# Dictionary
student = {
    "name": "Ram",
    "age": 22,
    "score": 87.5
}

print(type(age))
print(type(percentage))
print(type(student_name))
print(type(passed))
print(type(scores))
print(type(coordinates))
print(type(unique_scores))
print(type(student))
````

## 3. Using `type()`

The `type()` function tells us the data type of a value or variable.

````python
x = 100
y = 25.5
z = "Data Science"
a = False

print("x:", type(x))
print("y:", type(y))
print("z:", type(z))
print("a:", type(a))
````

## 4. Integers and Floats

Integers are whole numbers. Floats contain decimal values.

````python
age = 22
salary = 35000
average_score = 84.75

print("Age:", age)
print("Salary:", salary)
print("Average score:", average_score)

# Basic arithmetic
print("Salary + 5000 =", salary + 5000)
print("Salary - 5000 =", salary - 5000)
print("Salary * 2 =", salary * 2)
print("Salary / 2 =", salary / 2)
````

## 5. Strings

Strings are used to store text. They can be written using single or double quotes.

````python
first_name = "Prajapati"
last_name = "Ramdev"

full_name = first_name + " " + last_name

print("Full name:", full_name)
print("Name length:", len(full_name))
print("Uppercase:", full_name.upper())
print("Lowercase:", full_name.lower())
````

## 6. Boolean Values

A boolean has only two possible values: `True` or `False`.

Booleans are useful when making decisions, filtering data, and checking conditions.

````python
age = 22
passing_score = 40
score = 87

is_adult = age >= 18
has_passed = score >= passing_score

print("Is adult:", is_adult)
print("Has passed:", has_passed)
````

## 7. Lists

A list stores multiple values in an ordered collection.

Lists are **mutable**, meaning their contents can be changed.

````python
scores = [85, 90, 78, 92]

print("Scores:", scores)
print("First score:", scores[0])
print("Number of scores:", len(scores))

# Add a new score
scores.append(88)

print("Updated scores:", scores)
````

## 8. Tuples

A tuple is similar to a list, but tuples are **immutable** (their elements cannot normally be changed after creation).

````python
student_location = (19.07, 72.87)

print("Location:", student_location)
print("Latitude:", student_location[0])
print("Longitude:", student_location[1])
print("Type:", type(student_location))
````

## 9. Sets

A set stores **unique** values. Duplicate values are automatically removed.

````python
scores = {85, 90, 85, 78, 90, 92}

print("Unique scores:", scores)
print("Number of unique scores:", len(scores))
````

## 10. Dictionaries

A dictionary stores information as **key-value pairs**.

This is very useful for representing structured information such as a student's profile.

````python
student = {
    "name": "Ram",
    "age": 22,
    "course": "BCA",
    "score": 87.5
}

print("Student:", student)
print("Name:", student["name"])
print("Age:", student["age"])
print("Score:", student["score"])
````

## 11. Type Conversion

Type conversion means changing a value from one data type to another.

Common functions:
- `int()` → integer
- `float()` → decimal number
- `str()` → string
- `bool()` → boolean

A common real-world example is converting numbers received as text into numeric values.

````python
# String to integer
age_text = "22"
age_number = int(age_text)

# String to float
salary_text = "35000.50"
salary_number = float(salary_text)

# Number to string
score = 87
score_text = str(score)

print(age_number, type(age_number))
print(salary_number, type(salary_number))
print(score_text, type(score_text))
````

````python
# Why conversion matters

marks1 = "80"
marks2 = "90"

# Without conversion, + joins strings
print("As strings:", marks1 + marks2)

# Convert before mathematical calculation
total_marks = int(marks1) + int(marks2)

print("As numbers:", total_marks)
````

## 12. Basic Operations

Python supports arithmetic operations such as:
- Addition `+`
- Subtraction `-`
- Multiplication `*`
- Division `/`
- Floor division `//`
- Modulus `%`
- Power `**`


````python
a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** 2)
````

## 13. Real-World Data Science Example

Imagine we have information about an employee.

We can store age, salary, performance score, and employment status in Python variables.

````python
employee_name = "Rahul"
employee_age = 25
employee_salary = 45000.50
performance_score = 91.5
is_active = True

print("Employee:", employee_name)
print("Age:", employee_age)
print("Salary:", employee_salary)
print("Performance score:", performance_score)
print("Active employee:", is_active)

# Simple salary calculation
annual_salary = employee_salary * 12
print("Annual salary:", annual_salary)
````

## 14. Student Performance Example

Let's represent a student's marks and calculate the average.

````python
student_name = "Amit"
maths = 85
science = 90
python = 95

total = maths + science + python
average = total / 3

print("Student:", student_name)
print("Total:", total)
print("Average:", average)

passed = average >= 40
print("Passed:", passed)
````

# Interview Questions & Answers

### 1. What are the main built-in data types in Python?
Common built-in types include `int`, `float`, `str`, `bool`, `list`, `tuple`, `set`, and `dict`.

### 2. What is the difference between a list and a tuple?
Both are ordered collections. A list is mutable, while a tuple is immutable.

Example:
```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
```

### 3. Why is Python widely used in data science?
Python has simple syntax and a large ecosystem of data science libraries such as NumPy, pandas, Matplotlib, and scikit-learn. It is useful for data cleaning, analysis, visualization, and machine learning.

### 4. What does `type()` do?
`type()` tells you the data type of a value or variable.

Example:
```python
type(25)       # int
type(25.5)     # float
type("Python") # str
```

### 5. What is type conversion?
Type conversion changes a value from one data type to another, such as converting `"25"` into `25` using `int("25")`.
