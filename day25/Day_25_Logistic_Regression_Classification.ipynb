## Dataset
Breast Cancer Wisconsin Dataset from scikit-learn.


```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## 1. Load and Inspect Data


```python
data=load_breast_cancer()
X=pd.DataFrame(data.data,columns=data.feature_names)
y=pd.Series(data.target,name="target")
print("Shape:",X.shape)
print("Missing values:",X.isnull().sum().sum())
print(y.value_counts().sort_index())
X.head()
```

## 2. Train-Test Split
Stratification keeps class proportions similar in both sets.


```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
print("Train:",X_train.shape,"Test:",X_test.shape)
```

## 3. Feature Scaling
Fit the scaler only on training data to avoid data leakage.


```python
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

## 4. Train Logistic Regression


```python
model=LogisticRegression(max_iter=5000,random_state=42)
model.fit(X_train_scaled,y_train)
print("Model trained successfully.")
```

## 5. Predictions and Predicted Probabilities


```python
y_pred=model.predict(X_test_scaled)
y_proba=model.predict_proba(X_test_scaled)
probability_table=pd.DataFrame({"Actual":y_test.to_numpy(),"Predicted":y_pred,"P(malignant)":y_proba[:,0],"P(benign)":y_proba[:,1]})
print(probability_table.head(10))
```

## 6. Evaluation


```python
accuracy=accuracy_score(y_test,y_pred)
print(f"Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(classification_report(y_test,y_pred,target_names=["malignant","benign"]))
print("Confusion Matrix:")
print(confusion_matrix(y_test,y_pred))
```

## 7. Short Model Interpretation
- Logistic Regression estimates class probabilities for classification.
- `predict_proba()` returns the estimated probability for each class.
- Scaling helps when numerical features have different scales.
- Precision, recall and F1-score provide more detail than accuracy alone.
- Coefficients show how standardized features contribute to the model decision.

## 8. Interview Questions
**What is Logistic Regression?** A supervised learning algorithm commonly used for classification that estimates class probabilities.

**Why is it used for classification?** It is simple, interpretable, probability-based, and useful as a baseline model.

**What does `predict_proba()` return?** Estimated probability for each class for every sample.

**Why scale features?** To put numerical features on comparable scales and support stable optimization.
