# Day 28 — Decision Tree Classifier

**Level 2 · Day 28 · Data Science Track**

Train a Decision Tree classifier and explore how tree depth influences model performance.

## Objective
Understand tree-based classification and model complexity.

## Tools
Python, Pandas, Scikit-learn, Matplotlib, Jupyter Notebook

## Dataset
Iris Dataset from scikit-learn.


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report
```

## 1. Load and Inspect Dataset

```python
iris=load_iris()
X=pd.DataFrame(iris.data,columns=iris.feature_names)
y=pd.Series(iris.target,name='target')
print('Shape:',X.shape)
print('Classes:',iris.target_names)
print('Missing values:',X.isnull().sum().sum())
X.head()
```

## 2. Train-Test Split

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
print('Training samples:',len(X_train))
print('Testing samples:',len(X_test))
```

## 3. Train Decision Tree

```python
model=DecisionTreeClassifier(max_depth=3,random_state=42)
model.fit(X_train,y_train)
train_pred=model.predict(X_train)
test_pred=model.predict(X_test)
print('Training accuracy:',round(accuracy_score(y_train,train_pred),4))
print('Testing accuracy:',round(accuracy_score(y_test,test_pred),4))
```

## 4. Visualize the Tree

```python
plt.figure(figsize=(16,9))
plot_tree(model,feature_names=iris.feature_names,class_names=iris.target_names,filled=True,rounded=True)
plt.title('Decision Tree Classifier — max_depth=3')
plt.tight_layout()
plt.show()
```

## 5. Compare Different Tree Depths

```python
results=[]
for depth in range(1,11):
    m=DecisionTreeClassifier(max_depth=depth,random_state=42)
    m.fit(X_train,y_train)
    tr=accuracy_score(y_train,m.predict(X_train))
    te=accuracy_score(y_test,m.predict(X_test))
    results.append([depth,tr,te,tr-te])
results=pd.DataFrame(results,columns=['max_depth','train_accuracy','test_accuracy','accuracy_gap'])
results.round(4)
```

```python
plt.figure(figsize=(9,5))
plt.plot(results.max_depth,results.train_accuracy,marker='o',label='Training Accuracy')
plt.plot(results.max_depth,results.test_accuracy,marker='o',label='Testing Accuracy')
plt.xlabel('Maximum Tree Depth');plt.ylabel('Accuracy');plt.title('Performance Across Tree Depths')
plt.xticks(range(1,11));plt.grid(alpha=.3);plt.legend();plt.tight_layout();plt.show()
```

## 6. Best Tested Depth

```python
best_test=results.test_accuracy.max()
best_depth=int(results.loc[results.test_accuracy==best_test,'max_depth'].min())
print('Best tested max_depth:',best_depth)
print('Best testing accuracy:',round(best_test,4))
```

## 7. Feature Importance

```python
importance=pd.DataFrame({'feature':X.columns,'importance':model.feature_importances_}).sort_values('importance',ascending=False)
importance
```

```python
plt.figure(figsize=(9,5))
plt.barh(importance.feature,importance.importance)
plt.xlabel('Feature Importance');plt.title('Decision Tree Feature Importance');plt.gca().invert_yaxis();plt.tight_layout();plt.show()
```

## 8. Overfitting Analysis

A large gap between training and testing accuracy can indicate overfitting. Deeper trees are usually more complex and can memorize training-specific patterns.

**Accuracy gap = training accuracy − testing accuracy.**

## Interview Questions

**1. How does a decision tree make predictions?** It follows feature-based split rules from the root to a leaf and uses the leaf class as the prediction.

**2. What is overfitting?** Overfitting occurs when a tree learns training-specific noise and performs worse on unseen data.

**3. What does `max_depth` control?** It limits the maximum levels of the tree and therefore controls model complexity.
